Q5. Create Pandas DataFrame from dictionary and perform ranking, sorting and aggregation.

```python
import pandas as pd

data = {'Name': ['Asha', 'Ravi', 'Meena', 'John', 'Kiran', 'Sara'],
        'Dept': ['IT', 'HR', 'IT', 'HR', 'IT', 'HR'],
        'Salary': [50000, 40000, 60000, 45000, 55000, 40000]}
df = pd.DataFrame(data)

df['Rank'] = df['Salary'].rank(ascending=False)
print(df.sort_values('Salary', ascending=False))
print(df.groupby('Dept')['Salary'].agg(['mean', 'max', 'min']))
print(df['Salary'].sum())
```

Q6. Detect skewness and apply Min-Max scaling.

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

df = pd.read_csv('income.csv')
print(df['Income'].skew())

scaler = MinMaxScaler()
df['Income_scaled'] = scaler.fit_transform(df[['Income']])
print(df[['Income', 'Income_scaled']])
```

Q7. Create scatter plot and correlation heatmap.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('study.csv')

plt.scatter(df['Hours'], df['Marks'])
plt.xlabel('Study Hours')
plt.ylabel('Marks')
plt.title('Hours vs Marks')
plt.show()

sns.heatmap(df.corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

Q8. Train SVM with linear kernel and visualize decision boundary.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.svm import SVC

iris = load_iris()
X = iris.data[:100, 2:4]
y = iris.target[:100]

model = SVC(kernel='linear')
model.fit(X, y)
print(model.score(X, y))

w = model.coef_[0]
b = model.intercept_[0]
x_line = np.linspace(1, 5.2, 50)
y_line = -(w[0] * x_line + b) / w[1]

plt.scatter(X[:, 0], X[:, 1], c=y)
plt.plot(x_line, y_line, 'k')
plt.scatter(model.support_vectors_[:, 0], model.support_vectors_[:, 1], s=120, facecolors='none', edgecolors='k')
plt.ylim(0, 2.8)
plt.xlabel('Petal Length')
plt.ylabel('Petal Width')
plt.title('Linear SVM Decision Boundary')
plt.show()
```
