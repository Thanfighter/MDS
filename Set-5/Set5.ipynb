Q17. Create NumPy arrays and perform arithmetic and statistical operations using universal functions.

```python
import numpy as np

a = np.array([10, 20, 30, 40, 50])
b = np.array([1, 2, 3, 4, 5])

print(np.add(a, b))
print(np.subtract(a, b))
print(np.multiply(a, b))
print(np.divide(a, b))
print(np.power(b, 2))
print(np.mean(a), np.median(a), np.std(a))
print(np.min(a), np.max(a), np.sum(a))
print(np.var(a))
```

Q18. Import a CSV file, remove duplicate records, and calculate group-wise statistics using Pandas.

```python
import pandas as pd

df = pd.read_csv('orders.csv')
print(df.duplicated().sum())

df = df.drop_duplicates()
print(df.groupby('Category')['Amount'].agg(['count', 'mean', 'sum', 'max']))
```

Q19. Plot overlapping histograms to compare distributions of numerical features. Include a legend and suitable axis labels.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('scores.csv')

plt.hist(df['Maths'], bins=5, alpha=0.5, label='Maths')
plt.hist(df['English'], bins=5, alpha=0.5, label='English')
plt.xlabel('Marks')
plt.ylabel('Number of Students')
plt.title('Maths vs English Marks')
plt.legend()
plt.show()
```

Q20. Implement Logistic Regression for binary classification. Display the confusion matrix and ROC curve, and report an appropriate evaluation measure.

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, roc_curve, roc_auc_score

df = pd.read_csv('purchase.csv')

X = df[['Age', 'Salary']]
y = df['Buy']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3)

model = LogisticRegression()
model.fit(X_train, y_train)

pred = model.predict(X_test)
prob = model.predict_proba(X_test)[:, 1]

print("Accuracy:", accuracy_score(y_test, pred))
print("Confusion Matrix:\n", confusion_matrix(y_test, pred))
print("AUC:", roc_auc_score(y_test, prob))

fpr, tpr, _ = roc_curve(y_test, prob)
plt.plot(fpr, tpr)
plt.plot([0, 1], [0, 1], '--')
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve")
plt.show()
```
